```python
import pandas as pd
dados_pets = {
    'NomePet': ['Floco', 'Rex', 'Mia', 'Thor', 'Bolinha', 'Luna'],
    'Especie': ['Cachorro', 'Cachorro', 'Gato', 'Cachorro', 'Gato', 'Cachorro'],
    'Idade': [3, 8, 2, 12, 5, 1],
    'Peso_kg': [15, 25, 4, 30, 6, 8],
    'Vacinado': [True, True, False, True, True, False]}
df_pets = pd.DataFrame(dados_pets)
print("--- Clínica Veterinária ---")
print(df_pets)
print("\n --- 1. Animais da espécie Gato ---")
mascara_gatos = df_pets['Especie'] == 'Gato'
pets_gatos = df_pets[mascara_gatos]
print(pets_gatos)
print("\n--- 2. Filtro Numérico: Crie um DataFrame ---")
mascara_pesados = df_pets['Peso_kg'] > 20
pets_pesados = df_pets[mascara_pesados]
print(pets_pesados)
print("\n---3. Filtro Combinado (E): Crie um DataFrame pets_idosos_grandes que contenha 'Cachorro' E 'Idade' maior que 7.---")
mascara_cachorros = df_pets['Especie'] == 'Cachorro'
mascara_idades = df_pets['Idade'] > 7
mascara_combinada = (mascara_cachorros) & (mascara_idades)
pets_idosos_grandes = df_pets[mascara_combinada]
print(pets_idosos_grandes)
```

```python
#Exercício 2: Supermercado (Estoque)
import pandas as pd
dados_estoque = {
'Produto': ['Arroz', 'Feijão', 'Óleo', 'Sal', 'Macarrão', 'Café'],
'Categoria': ['Grãos', 'Grãos', 'Mercearia', 'Mercearia', 'Massas', 'Mercearia'],
'Qtd_Estoque': [50, 15, 10, 100, 40, 5],
'Preco_Unit': [25.00, 8.00, 7.50, 3.00, 6.00, 15.00]
}

#1. Filtro Simples: Crie um DataFrame estoque_baixo que mostre apenasprodutos onde a 'Qtd_Estoque' é menor ou igual a 10.
df_estoque = pd.DataFrame(dados_estoque)
print("\n--- Controle de Estoque ---")
print(df_estoque)
print("\n--- 1. Produtos com Estoque Baixo (Qtd <=10)' ---")
mascara_Qtd_Estoque = df_estoque['Qtd_Estoque']  <= 10
estoque_baixo = df_estoque[mascara_Qtd_Estoque]
print(estoque_baixo)

#2. Filtro Combinado (OU): Crie um DataFrame reposicao_urgente que mostreprodutos da categoria 'Grãos' OU da categoria 'Mercearia'.
condicao_grao = df_estoque['Categoria'] == 'Grãos'
condicao_mercearia = df_estoque['Categoria'] == 'Mercearia'
mascara_final = (condicao_grao) | (condicao_mercearia)
reposicao_urgente = df_estoque[mascara_final]

print("\n--- 2. Produtos da Categoria 'Grãos' OU 'Mercearia' ---")
print(reposicao_urgente)

#3. Filtro Combinado (E): Crie um DataFrame oferta_graos que mostre produtos da categoria 'Grãos' E que tenham 'Preco_Unit' menor que 10.00.
oferta_graos = df_estoque['Categoria'] == 'Grãos'
preco_unitario = df_estoque['Preco_Unit'] < 10.00
condicao_final = (condicao_grao) & (preco_unitario)
oferta_graos = df_estoque[condicao_final]
print("\n--- 3. Produtos da Categoria 'Grãos' com Preço Unitário < 10.00 ---")
print(oferta_graos)
                          
```

```python
#Exercício 3: Desempenho Acadêmico
#Crie o DataFrame df_alunos e filtre os alunos com base em seu desempenho e frequência.
import pandas as pd
dados_alunos = {
'Matricula': [101, 102, 103, 104, 105, 106],
'Nome': ['Ana', 'Bruno', 'Carla', 'Daniel', 'Elisa', 'Fábio'],
'Nota_Final': [8.5, 6.0, 9.5, 4.0, 7.0, 5.5],
'Faltas': [2, 5, 0, 10, 1, 3]
}
df_alunos = pd.DataFrame(dados_alunos)
print("\n--- Desempenho Acadêmico ---")
print(df_alunos)

#1. Filtro de Aprovação (Conceitual): Usando a lógica do exemplo , crie um DataFrame aprovados que contenha apenas alunos com 'Nota_Final' maior ou igual a 7.0.
mascara_aprovacao = df_alunos['Nota_Final'] >= 7.0
aprovados = df_alunos[mascara_aprovacao]   
print(aprovados)
#2. Filtro de Reprovação por Falta: Crie um DataFrame reprovados_falta que contenha alunos com 'Faltas' maior que 5.
mascara_reprovados = df_alunos['Faltas'] > 5
reprovaddos_falta = df_alunos[mascara_reprovados]
print(reprovaddos_falta)
#3. Filtro de Aluno Exemplar (E): Crie um DataFrame alunos_exemplares que contenha alunos com 'Nota_Final' maior que 9.0 E 'Faltas' igual a 0.
condicao_nota = df_alunos['Nota_Final'] > 9.0
condicao_faltas = df_alunos['Faltas'] == 0
condicao_exemplar = (condicao_nota) & (condicao_faltas)
alunos_exemplares = df_alunos[condicao_exemplar]
print(alunos_exemplares)
```

```python
#Exercício 4: Vendas da Equipe
#Crie o DataFrame df_vendas e filtre os dados para analisar o desempenho dos vendedores.

import pandas as pd
dados_vendas = {
'Vendedor': ['Rafael', 'Bia', 'Carlos', 'Rafael', 'Bia', 'Carlos'],
'Regiao': ['Sul', 'Sudeste', 'Nordeste', 'Sudeste', 'Sul', 'Sudeste'],
'Valor_Venda': [1200, 1500, 800, 2000, 1800, 1300],
'Tipo': ['Online', 'Presencial', 'Online', 'Online', 'Presencial', 'Presencial']
}
df_vendas = pd.DataFrame(dados_vendas)
print("\n--- Vendas da Equipe ---")
print(df_vendas)

#1. Filtro de Vendedor: Crie um DataFrame vendas_bia que mostre apenas as vendas feitas pela vendedora 'Bia'.

print("\n--- 1. Vendas da Vendedora 'Bia' ---")
mascara_bia = df_vendas['Vendedor'] == 'Bia'
vendas_bia = df_vendas[mascara_bia]
print(vendas_bia)

#2. Filtro de Alto Valor: Crie um DataFrame vendas_altas que mostre apenas vendas com 'Valor_Venda' maior que 1400.
print("\n--- 2. Vendas com Valor Maior que 1400 ---")
mascara_vendas_altas = df_vendas['Valor_Venda'] > 1400
vendas_altas = df_vendas[mascara_vendas_altas]
print(vendas_altas)

#3. Filtro Combinado (OU): Crie um DataFrame vendas_foco que mostre vendas da Região 'Sul' OU da Região 'Sudeste'.
condicao_sul = df_vendas['Regiao'] == 'Sul'
condicao_sudeste = df_vendas['Regiao'] == 'Sudeste'
mascara_ou = (condicao_sul) | (condicao_sudeste)
vendas_foco = df_vendas[mascara_ou]
print("\n--- 3. Vendas na Região 'Sul' OU 'Sudeste' ---")
print(vendas_foco)
```
