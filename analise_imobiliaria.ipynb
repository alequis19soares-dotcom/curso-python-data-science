```python
#2. Preparando o Ambiente - EXEMPLO
#Objetivo: Aplicar os conceitos de Histograma e Gráfico de Dispersão criar um novo conjunto de dados. Você usará as bibliotecas pandas, matplotlib e seaborn para investigar a distribuição de preços de imóveis e a relação entre o tamanho e o preço
#Contexto:
#Imagine que você recebeu um pequeno conjunto de dados sobre imóveis em uma determinada região. O conjunto de dados contém o tamanho do imóvel em metros quadrados e o seu preço de venda.
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
dados_imoveis = pd.DataFrame({
'tamanho_m2': [40.0, 55.0, 80.0, 100.0, 120.0],
'preco': [200, 290, 350, 500, 600]
})

# Criando um histograma para Análise Imobiliaria
plt.figure(figsize=(9, 5)) # Define o tamanho da figura
sns.histplot(dados_imoveis, x='preco', kde=True) # kde=True desenha uma linha de densidade
plt.title('Distribuição de Preços dos Imóveis')
plt.xlabel('Preço dos Imóveis(em mil R$)')
plt.ylabel('Contagem (Frequência)')# Eixo Y de um Histograma é sempre a Contagem
plt.show() # Exibe o gráfico
```

```python
# Criando um gráfico de dispersão
plt.figure(figsize=(9, 5))
sns.scatterplot(data=dados_imoveis, x='tamanho_m2', y='preco', s=150)
plt.title('Relação entre Tamanho e Preços dos Imóveis')
plt.xlabel('tamanho (m$^2$)')
plt.ylabel('Preço (em mil R$)')
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(9, 5))
sns.regplot(x='tamanho_m2', y='preco', data=dados_imoveis)
plt.title('Linha de Regressão: Tamanho vs. Preço')
plt.xlabel('Tamanho do Imóvel em Metros')
plt.ylabel('Preço dos Imóveis (em mil R$)')
plt.grid(True)
plt.show()
# Para ver o número da correlação:
correlacao = dados_imoveis.corr()
print("\nMatriz de Correlação:")
print(correlacao)
```
