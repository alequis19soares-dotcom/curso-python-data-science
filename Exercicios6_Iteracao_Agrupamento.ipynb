```python
import pandas as pd

# --- Amostra 1: Desempenho de Alunos ---
dados_alunos = {
'aluno': ['Ana', 'Bruno', 'Carla', 'David', 'Elisa', 'Fábio', 'Gabriela', 'Helio'],
'turma': ['A', 'B', 'A', 'B', 'A', 'B', 'A', 'B'],
'nota_final': [8.5, 7.0, 9.0, 5.5, 9.5, 6.5, 7.5, 8.0],
'faltas': [2, 5, 1, 6, 0, 4, 3, 2]
}
df_alunos = pd.DataFrame(dados_alunos)
print("--- Amostra 1: Desempenho de Alunos ---")
print(df_alunos)

# 1. Laço for (Conceitual)
lista_alunos = df_alunos['aluno'].tolist()
print("\n--- 1. Laço for (Boas-vindas) ---")
for nome in lista_alunos:
    print(f"Boas-vindas, {nome}!")

# 2. Laço for (com range)
print("\n--- 2. Laço for (Top 3) ---")
for rank in range(1, 4):
    print(f"Rank {rank}")

# 3. groupby (Média)
media_por_turma = df_alunos.groupby('turma')['nota_final'].mean()
print("\n--- 3. Média da Nota Final por Turma ---")
print(media_por_turma)
melhor_turma = media_por_turma.idxmax()
melhor_nota = media_por_turma.max()
print(f"\nMelhor desempenho médio: Turma {melhor_turma} com nota {melhor_nota:.2f}")

# 4. groupby (Soma)
total_faltas_por_turma = df_alunos.groupby('turma')['faltas'].sum()
print("\n--- 4. Total de Faltas por Turma ---")
print(total_faltas_por_turma)

# 5. groupby (Contagem)
contagem_alunos_por_turma = df_alunos.groupby('turma')['aluno'].size()
print("\n--- 5. Contagem de Alunos por Turma ---")
print(contagem_alunos_por_turma)
```

```python
#Amostra 2: Pedidos de Restaurante
# DataFrame simulando pedidos em um restaurante, com pratos, categorias e preços.

import pandas as pd
# --- Amostra 2: Pedidos de Restaurante ---
dados_pedidos = {
'mesa': [1, 2, 1, 3, 2, 1, 3, 2],
'categoria': ['Principal', 'Bebida', 'Sobremesa', 'Principal', 'Bebida', 'Principal', 'Sobremesa', 'Principal'],
'item': ['Carbonara', 'Refrigerante', 'Tiramisu', 'Lasanha', 'Água', 'Pizza', 'Pudim', 'Steak'],
'preco': [45.00, 8.00, 22.00, 50.00, 6.00, 48.00, 18.00, 65.00]
}
df_pedidos = pd.DataFrame(dados_pedidos)
print("\n--- Amostra 2: Pedidos de Restaurante ---")
print(df_pedidos)

#Exercícios (Amostra 2):
#1. Laço for (Conceitual): Crie uma lista com os item pedidos. Use um laço for para imprimir cada item, formatado como "Preparando: Carbonara".
itens_pedidos = df_pedidos['item'].tolist()
print("\n--- 1. Laço for (Preparando) ---")
for nome in itens_pedidos:
    print(f"Preparando, {nome}!")

#2. groupby (Soma): Qual foi o valor total faturado (soma do preco) por categoria de item (Principal, Bebida, Sobremesa)?
total_faturado = df_pedidos.groupby('categoria')['preco'].sum()
print("\n--- 2. Total Fatura por Categoria ---")
print(total_faturado)

#3. groupby (Média): Qual o preço médio dos itens em cada categoria?
media_por_item = df_pedidos.groupby('categoria')['preco'].mean()
print("\n--- 3. Média de Preço por Categoria ---")
print(media_por_item)
melhor_item = media_por_item.idxmax()
melhor_preco = media_por_item.max()
print(f"\nMelhor preço_médio: Categoria {melhor_item} com nota {melhor_preco:.2f}")

#4. groupby (Soma): Qual foi o gasto total (soma do preco) por mesa?
total_gasto_mesa = df_pedidos.groupby('mesa')['preco'].sum()
print("\n--- 4. Total Gasto por Mesa ---")
print(total_gasto_mesa)

#5. groupby (Múltiplas Funções): Usando .agg(), mostre o faturamento sum (Total) e a count (No de Pedidos)para cada categoria.
resultado_agg = df_pedidos.groupby('categoria')['preco'].agg(['sum', 'count'])
print("\n --- 5.Agregação Mútipla (Total e Contagem por Categoria) ---")
print(resultado_agg)
```

```python
#Amostra 3: Inventário de E-commerce DataFrame de inventário de produtos, com departamentos e quantidade em estoque.

import pandas as pd
# --- Amostra 3: Inventário de E-commerce ---
dados_inventario = {
'departamento': ['Eletrônicos', 'Livros', 'Eletrônicos', 'Roupas', 'Livros', 'Eletrônicos', 'Roupas', 'Roupas'],
'item': ['Fone de Ouvido', 'Dom Casmurro', 'Mouse', 'Camiseta', 'O Gene', 'Teclado', 'Calça Jeans', 'Meias'],
'estoque': [50, 120, 80, 200, 90, 60, 150, 300],
'preco_unitario': [150.00, 40.00, 80.00, 60.00, 55.00, 220.00, 120.00, 25.00]
}
df_inventario = pd.DataFrame(dados_inventario)
print("\n--- Amostra 3: Inventário de E-commerce ---")
print(df_inventario)

#Exercícios (Amostra 3):
#1. Laço for (Conceitual): Obtenha uma lista de departamentos únicos (ex:df_inventario['departamento'].unique()). Use um laço for para imprimir uma mensagem como "Verificando inventário do departamento: Eletrônicos".
departamentos_unicos = df_inventario['departamento'].unique()
print("Departamentos Únicos Encontrados:", departamentos_unicos)
print("\n ---1. Verificação de Inventário por Departamento ---")
for dep in departamentos_unicos:
    print(f" Verificando Inventário do departamento: {dep}")

#2. groupby (Soma): Qual é a quantidade total de itens em estoque por departamento?estoque
estoque_por_departamento = df_inventario.groupby('departamento')['estoque'].sum()
print('--- Quantidade Total de Itens em Estoque por Departamento ---')
print(estoque_por_departamento)

#3. groupby (Média): Qual é o preço unitário médio dos produtos em cada departamento?
preco_medio_por_departamento = df_inventario.groupby('departamento')['preco_unitario'].mean()

# Aplicamos o arredondamento para 2 casas decimais, para facilitar a leitura
preco_medio_formatado = preco_medio_por_departamento.round(decimals=2)

print("\n--- 3. Preço Unitário Médio por Departamento (groupby Média) ---")
print(preco_medio_formatado)

#4. groupby (Contagem): Quantos tipos diferentes de produtos (linhas) existem por departamento?
contagem_produtos_por_departamento = df_inventario.groupby('departamento')['item'].nunique()
print("\n--- 4. Contagem de Produtos por Departamento (groupby Contagem) ---")
print(contagem_produtos_por_departamento)

#5. Desafio groupby: Primeiro, crie uma nova coluna no DataFrame chamada valor_total_estoque (que éestoque * preco_unitario). Depois, use groupby para calcular o valor total (soma da nova coluna) do estoque por departamento.
df_inventario['valor_total_estoque'] = df_inventario['estoque'] * df_inventario['preco_unitario']
valor_total_por_departamento = df_inventario.groupby('departamento')['valor_total_estoque'].sum()
print("\n--- 5. Valor Total do Estoque por Departamento (Desafio groupby) ---")
print(valor_total_por_departamento)
```
