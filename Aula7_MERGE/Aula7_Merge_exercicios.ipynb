```python
# Importa especificamente a classe 'date' de dentro do módulo 'datetime'. # # O 'datetime' é o módulo padrão do
#Python usado para manipular datas e horas. # Ao importar 'date' diretamente (em vez de importar o módulo
#'datetime' inteiro), # podemos criar um objeto de data (ano, mês, dia) de forma mais curta, # usando apenas
#'date(2025, 11, 5)' em vez de 'datetime.date(2025, 11, 5)'.

#Código Python

import pandas as pd
from datetime import date
# --- DataFrame 1: Funcionários ---
data_funcionarios = {
"ID_Funcionario": [101, 102, 103, 104, 105, 106, 107, 108],
"Nome": ["Ana", "Bruno", "Carla", "Daniel", "Elisa", "Fábio", "Gabriela", "Hugo"],
"Sobrenome": ["Silva", "Costa", "Oliveira", "Souza", "Pereira", "Martins", "Alves", "Ribeiro"],
"ID_Departamento": [1, 2, 1, 3, 2, 1, 3, 2],
"ID_Cargo": [10, 11, 12, 13, 11, 10, 14, 12],
"Data_Contratacao": pd.to_datetime(["2022-01-15", "2021-05-30", "2022-03-10", "2020-11-01", "2023-07-20",
"2022-02-18", "2021-09-05", "2023-11-01"])
}
df_funcionarios = pd.DataFrame(data_funcionarios)
# --- DataFrame 2: Departamentos ---
data_departamentos = {
"ID_Departamento": [1, 2, 3, 4],
"Nome_Departamento": ["Engenharia", "Marketing", "RH", "Financeiro"],
"Localizacao": ["Andar 10", "Andar 8", "Andar 9", "Andar 9"]
}
df_departamentos = pd.DataFrame(data_departamentos)
# --- DataFrame 3: Cargos e Salários ---
data_cargos_salarios = {
"ID_Cargo": [10, 11, 12, 13, 14, 15],
"Nome_Cargo": ["Eng. Software Jr", "Analista de Marketing", "Eng. Software Sr", "Gerente de Contas",
"Recrutador", "Analista Financeiro"],
"Salario_Base": [5500, 6000, 9500, 8000, 5000, 7000],
"Nivel": ["Junior", "Pleno", "Senior", "Pleno", "Junior", "Pleno"]
}
df_cargos_salarios = pd.DataFrame(data_cargos_salarios)
# --- DataFrame 4: Projetos ---
data_projetos = {
"ID_Projeto": ["P-1001", "P-1002", "P-1003"],
"Nome_Projeto": ["Sistema de Login", "Campanha de Natal", "Otimização de Performance"],
"ID_Gerente_Projeto": [103, 104, 103],
"Status": ["Em Andamento", "Planejamento", "Concluído"]
}
df_projetos = pd.DataFrame(data_projetos)
# --- DataFrame 5: Alocação de Horas ---
data_alocacao = {
"ID_Alocacao": [1, 2, 3, 4, 5, 6, 7],
"ID_Funcionario": [101, 103, 107, 102, 105, 101, 108],
"ID_Projeto": ["P-1001", "P-1001", "P-1003", "P-1002", "P-1002", "P-1003", "P-1002"],
"Horas_Semanais_Alocadas": [20, 40, 30, 25, 15, 10, 40]
}
df_alocacao = pd.DataFrame(data_alocacao)

#Exercício 1: Funcionários e seus Departamentos Objetivo: Crie um novo DataFrame df_func_depto que combine df_funcionarios e df_departamentos para mostrar o nome e a localização do departamento de cada funcionário.
#Dica: Use pd.merge() com a coluna-chave on="ID_Departamento".
df_func_depto = pd.merge(df_funcionarios, df_departamentos, on="ID_Departamento")
print("--- 1. Funcionários e seus Departamentos (merge)---")
print(df_func_depto[["Nome", "Sobrenome", "Nome_Departamento", "Localizacao"]])
print("\n")

#Exercício 2: Informações Salariais dos Funcionários Objetivo: Crie um novo DataFrame df_func_salario que combine df_funcionarios e df_cargos_salarios para mostrar o nome do cargo, o salário base e o nível (Junior, Pleno, Sênior) de cada funcionário.
#Dica: A coluna-chave para esta junção é on="ID_Cargo".
df_func_salario = pd.merge(df_funcionarios, df_cargos_salarios, on="ID_Cargo")
print("---2. Informações Salariais dos Funcionários (merge)---")
print(df_func_salario[['Nome', 'Sobrenome', 'Nome_Cargo', 'Salario_Base', 'Nivel']])
print("\n")

#Exercício 3: Relatório Completo de Alocação de Projetos Objetivo: Crie um relatório df_alocacao_completa que mostre quem está trabalhando em qual projeto. O DataFrame final deve incluir o Nome do Funcionário (de df_funcionarios) e o Nome do Projeto (de df_projetos).df_projetos).
#Dica: Este exercício requer dois merges separados, encadeados:
#1. Junte df_alocacao com df_funcionarios usando on="ID_Funcionario".
#2. Junte o resultado anterior com df_projetos usando on="ID_Projeto".
df_alocacao_completa = (
    df_alocacao.merge(
        df_funcionarios[['ID_Funcionario', 'Nome', 'Sobrenome']], 
        
        on ="ID_Funcionario"
        ) 
        .merge(
            df_projetos[['ID_Projeto', 'Nome_Projeto']], 
            on="ID_Projeto")
)
print("---3. Relatório Completo de Alocação de Projetos (Merge Encadeado)---")
print(df_alocacao_completa[['Nome', 'Sobrenome', 'Nome_Projeto', ]])
print("\n")

#Exercício 4: Média Salarial por Departamento
#Objetivo: Calcule o salário base médio para cada departamento. Pré-requisito: Você precisará de um DataFrame que contenha tanto o Nome_Departamento quanto o Salario_Base. Use os merges dos Exercícios 1 e 2 (ou crie um novo) para juntar df_funcionarios, df_departamentos e df_cargos_salarios.
#Dica: Use .groupby("Nome_Departamento")["Salario_Base"].mean() no DataFrame combinado.
#Primiero Merge: Adiciona o nome do Departamento.

df_combinado = pd.merge(
    df_funcionarios, 
    df_departamentos, 
    on="ID_Departamento", #busca a coluna em comum
    how='left'            # Mantém o resultado anterior (todos os funcionários)   
    )
print("---4. DataFrame Combinado (Funcionários + Departamentos) ---")
print(df_combinado)
df_combinado = pd.merge(
    df_combinado,
    df_cargos_salarios,
    on='ID_Cargo',   # Coluna em comum para trazer os salários
    how='left'       # Mantém o resultado anterior (todos os funcionários com departamentos)
)
print("\n")

print("\n--- Dataframe Combinado Final (Funcionários + Departamentos + Cargos e Salários) ---")
print(df_combinado)
# Agora, calcula a média salarial por departamento
media_salarial_por_departamento = df_combinado.groupby("Nome_Departamento")["Salario_Base"].mean()
print("---4. Média Salarial por DepartamentO")
print(media_salarial_por_departamento)
print("\n")
#Exercício 5: Contagem de Funcionários por Nível
#Objetivo: Conte quantos funcionários existem em cada nível (Junior, Pleno, Senior). Pré-requisito: Você precisará do DataFrame df_func_salario do Exercício 2, que contém a coluna "Nivel".
#Dica: Use .groupby("Nivel") e, em seguida, .size() ou ["ID_Funcionario"].count() para contar as ocorrências.
```
