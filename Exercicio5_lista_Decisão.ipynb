```python
# ==============================
# CONFIGURAÇÃO PADRÃO PARA EXIBIÇÃO COMPLETA
# ==============================
import pandas as pd
from IPython.display import display

# Exibir todas as linhas e colunas dos DataFrames
pd.set_option('display.max_rows', None)
pd.set_option('display.max_columns', None)

# Ajustar largura máxima da tela (evita quebra de linha)
pd.set_option('display.width', 1000)

# Mostrar até 200 caracteres por coluna (texto longo)
pd.set_option('display.max_colwidth', 200)

# Exibir valores numéricos sem notação científica
pd.set_option('display.float_format', '{:,.2f}'.format)

print("✅ Configuração de exibição ajustada para mostrar tudo sem truncar!\n")

dados_pets = {
'NomePet': ['Floco', 'Rex', 'Mia', 'Thor', 'Bolinha', 'Luna'],
'Especie': ['Cachorro', 'Cachorro', 'Gato', 'Cachorro', 'Gato', 'Cachorro'],
'Idade': [3, 8, 2, 12, 5, 1],
'Peso_kg': [15, 25, 4, 30, 6, 8],
'Vacinado': [True, True, False, True, True, False]
}
df_pets = pd.DataFrame(dados_pets)
print("--- Clínica Veterinária ---")
print(df_pets)

print("\n --- 1. Animais da espécie Gato ---")
mascara_gatos = df_pets['Especie'] == 'Gato'
pets_gatos = df_pets[mascara_gatos]
print(pets_gatos)

print("\n--- 2. Filtro Numérico: Crie um DataFrame ---")
mascara_pesados = df_pets['Peso_kg'] > 20
pets_pesados = df_pets[mascara_pesados]
print(pets_pesados)

print("\n---3. Filtro Combinado (E): Crie um DataFrame pets_idosos_grandes que contenha 'Cachorro' E 'Idade' maior que 7.---")
mascara_cachorros = df_pets['Especie'] == 'Cachorro'
mascara_idades = df_pets['Idade'] > 7
mascara_combinada = (mascara_cachorros) & (mascara_idades)
pets_idosos_grandes = df_pets[mascara_combinada]
print(pets_idosos_grandes)
```

```python
#Exercício 2: Supermercado (Estoque)
#Crie o DataFrame df_estoque e filtre os produtos que precisam de reposição ou estão em promoção.
import pandas as pd
dados_estoque = {
'Produto': ['Arroz', 'Feijão', 'Óleo', 'Sal', 'Macarrão', 'Café'],
'Categoria': ['Grãos', 'Grãos', 'Mercearia', 'Mercearia', 'Massas', 'Mercearia'],
'Qtd_Estoque': [50, 15, 10, 100, 40, 5],
'Preco_Unit': [25.00, 8.00, 7.50, 3.00, 6.00, 15.00]
}
df_estoque = pd.DataFrame(dados_estoque)
print("\n--- Exercicio 2 - Controle de Estoque (Supermercado) ---")
print(df_estoque)

print("\n --- 1 - Filtro Simples: estoque_baixo (Qtd_Estoque <= 10 ---")
estoque_baixo = df_estoque[df_estoque['Qtd_Estoque'] <= 10]
print(estoque_baixo)


print("\n--- 2. Filtro Combinado (OU): reposicao_urgente ('Grãos' OU 'Mercearia'---")
condicao_graos = df_estoque['Categoria'] == 'Grãos'
condicao_mercearia = df_estoque['Categoria'] == 'Mercearia'
mascara_final = (condicao_graos) | (condicao_mercearia)
reposicao_urgente = df_estoque[mascara_final]
print(reposicao_urgente)

print("\n --- 3. Filtro Combinado (E) oferta_graos (categoria 'Grãos' E que tenham 'Preco_Unit' < 10.00.")
condicao_graos = df_estoque['Categoria']  == 'Grãos'
condicao_preco = df_estoque['Preco_Unit'] <10
oferta_graos = df_estoque[condicao_graos & condicao_preco]
print(oferta_graos)
```

```python
#Exercício 3: Desempenho Acadêmico
#Crie o DataFrame df_alunos e filtre os alunos com base em seu desempenho e frequência.
import pandas as pd
dados_alunos = {
'Matricula': [101, 102, 103, 104, 105, 106],
'Nome': ['Ana', 'Bruno', 'Carla', 'Daniel', 'Elisa', 'Fábio'],
'Nota_Final': [8.5, 6.0, 9.5, 4.0, 7.0, 5.5],
'Faltas': [2, 5, 0, 10, 1, 3]
}
df_alunos = pd.DataFrame(dados_alunos)
print("\n---Exercício 3 -  Desempenho Acadêmico ---")
print(df_alunos)

print("\n 1. Filtro de Aprovação (Conceitual): 'Nota_Final' maior ou igual a 7.0.")
aprovados = df_alunos[df_alunos['Nota_Final'] >= 7.0]
print('\n --- 1. Alunos Aprovados (Nota >= 7.0) ---')
print(aprovados)

#2. Filtro de Reprovação por Falta: 'Faltas' maior que 5.
reprovados_falta = df_alunos[df_alunos['Faltas'] > 5]
print("\n--- 2. Reprovados por Faltas (> 5 faltas) ---")
print(reprovados_falta)
#3. Filtro de Aluno Exemplar (E): 'Nota_Final' maior que 9.0 E 'Faltas' = 0.
condicao_nota = df_alunos['Nota_Final'] > 9.0
condicao_faltas = df_alunos['Faltas'] == 0
alunos_exemplares = df_alunos[condicao_nota & condicao_faltas]
print("\n---3. Alunos Exemplares ---")
print(alunos_exemplares)
```

```python
#Exercício 4: Vendas da Equipe
#Crie o DataFrame df_vendas e filtre os dados para analisar o desempenho dosvendedores.
import pandas as pd
dados_vendas = {
'Vendedor': ['Rafael', 'Bia', 'Carlos', 'Rafael', 'Bia', 'Carlos'],
'Regiao': ['Sul', 'Sudeste', 'Nordeste', 'Sudeste', 'Sul', 'Sudeste'],
'Valor_Venda': [1200, 1500, 800, 2000, 1800, 1300],
'Tipo': ['Online', 'Presencial', 'Online', 'Online', 'Presencial', 'Presencial']
}
df_vendas = pd.DataFrame(dados_vendas)
print("\n--- Exercício 4. Vendas da Equipe ---")
print(df_vendas)

#1. Filtro de Vendedor: Vendedor = 'Bia'
vendas_bia = df_vendas[df_vendas['Vendedor'] == 'Bia']
print("\n --- 1. Vendas da Vendedora Bia ---")
print(vendas_bia)

#2. Filtro de Alto Valor: 'Valor_Venda' maior que 1400.
vendas_altas = df_vendas[df_vendas['Valor_Venda'] > 1400]
print("\n--- 2. Vendas Alto Valor(> 1400) ---")
print(vendas_altas)

#3. Filtro Combinado (OU): Região 'Sul' OU da Região 'Sudeste'.
condicao_sul = df_vendas['Regiao'] == 'Sul'
condicao_sudeste = df_vendas['Regiao'] == 'Sudeste'
vendas_foco = df_vendas [(condicao_sul | condicao_sudeste)]
print("\n --- 3.Vendas foco (Sul OU Sudeste) --- ")
print(vendas_foco)

#4. Filtro Diferente de (!=): Tipo != 'Online'. 
vendas_nao_online = df_vendas[df_vendas['Tipo'] != 'Online']
print("\n--- 4. Vendas Não Online ---")
print(vendas_nao_online)
```

```python
#Exercício 5: Imobiliária
#Crie o DataFrame df_imoveis e filtre os anúncios que correspondem aos critérios dos clientes.
import pandas as pd
dados_imoveis = {
'Bairro': ['Centro', 'Praia', 'Centro', 'Norte', 'Praia', 'Sul'],
'Tipo': ['Apartamento', 'Casa', 'Apartamento', 'Apartamento', 'Casa', 'Casa'],
'Area_m²': [60, 120, 45, 80, 150, 90],
'Preco_Aluguel': [1500, 2800, 1200, 1600, 3500, 2000],
'Quartos': [2, 3, 1, 2, 4, 3]
}
df_imoveis = pd.DataFrame(dados_imoveis)
print("\n--- Imobiliária ---")
print(df_imoveis)

#Tarefas (Filtros):

#1. Filtro Simples (Tipo): 'Tipo' 'Casa'.
apenas_casas = df_imoveis[df_imoveis['Tipo'] == 'Casa']
print("\n--- 1. Apenas Casas ---")
print(apenas_casas)

#2. Filtro de Custo (E): 'Preco_Aluguel' < 2000 E localizados no 'Centro'.
condicao_preco_acessivel = df_imoveis['Preco_Aluguel'] <2000
condicao_centro = df_imoveis['Bairro'] == 'Centro'
#Combinação E (&)
aluguel_acessivel = df_imoveis[condicao_preco_acessivel & condicao_centro]

print("\n --- 2. Aluguel Cessível (Centro E <2000) ---")
print(aluguel_acessivel)

#3. Filtro de Cliente Exigente (E):  'Praia' , que sejam 'Casa' , E que tenham 'Area_m2' > 130.
condicao_praia = df_imoveis['Bairro'] == 'Praia'
condicao_casa = df_imoveis['Tipo'] == 'Casa'
condicao_area = df_imoveis ['Area_m²'] > 130

#Combinação E (&) Tripla
cliente_praia = df_imoveis[condicao_praia & condicao_casa & condicao_area]
print("\n --- 3. Cliente Praia (Praia E Casa E Área > 130) --- ")
print(cliente_praia)

#4. Filtro de Opções (OU):  Imovéis com 1 quarto OU 2 quartos.
condicao_um_quarto = df_imoveis['Quartos'] == 1
condicao_dois_quartos = df_imoveis['Quartos'] == 2

#Combinação OU (|) - Obrigatório usar parênteses () em cada condição.
opcoes_pequenas = df_imoveis[ (condicao_um_quarto) | (condicao_dois_quartos)]

print("\n --- 4. Opções Pequenas (1 Quarto OU 2 Quartos) ---")
print(opcoes_pequenas)
```
