```python
# AULA 1 - EXERCÍCIO
import pandas as pd
#Dados de preços de imóveis em um bairro (em milhares de R$)
#Note a mansão de R$ 8 milhoes que é um outlier.
dados_imoveis =pd.DataFrame({
    'tipo': ['Apartamento', 'Casa', 'Apartamento', 'Casa','Studio', 'Casa', 'Mansão'],
    'preco_milhar': [300,450,350,400,250,500,8000]

})

print("\n--- Preços dos Imóveis no Bairro (em milhares de R$)---")
print(dados_imoveis)

#1. Calcule a média dos preços.
media_precos= dados_imoveis['preco_milhar'].mean()
print (f"O preço MÉDIA dos imóveis é: {media_precos:.2f}")

#2 Calcule a mediana dos preços.
mediana_precos = dados_imoveis['preco_milhar'].median()
print(f"O preço MEDIANA dos imóveis é:{mediana_precos:.2f}")

#3.Responda à pergunta do cenário
#Responda: Qual valor você apresentaria ao seu cliente como o "preço típico" e por quê?
print("\n Conclusão: Preço mediano dos imóveis é: 400.000 e esse seria o preço tipico de casas no determinado bairro, levando em conta que a media dos valores ficou elevada, porque temos uma mansão com preço elevado")
```

```python
# =============================================================================
# AULA 2 - CÓDIGO 1: CALCULANDO MEDIDAS DE DISPERSÃO
# =============================================================================
import pandas as pd
# Reutilizando nosso DataFrame da aula anterior.
dados_notas = pd.DataFrame({
'aluno': ['Ana', 'Bruno', 'Carla', 'Daniel', 'Elisa', 'Fábio', 'Gabi', 'Hugo', 'Inês', 'João'],
'nota_final': [8.5, 7.0, 9.0, 5.5, 10.0, 7.0, 6.5, 8.0, 9.5, 4.0]
})
print("--- Tabela de Notas dos Alunos ---")
print(dados_notas)
print("\n" + "="*40 + "\n")

# --- Calculando a AMPLITUDE ---
# É simplesmente a diferença entre o valor máximo e o mínimo.

valor_maximo = dados_notas['nota_final'].max()
valor_minimo = dados_notas['nota_final'].min()
amplitude = valor_maximo - valor_minimo
print(f"A nota máxima foi: {valor_maximo}")
print(f"A nota mínima foi: {valor_minimo}")
print(f"A AMPLITUDE das notas é: {amplitude:.2f}")
print("\n" + "-"*40 +"\n")

# --- Calculando a VARIÂNCIA ---
# O método .var() calcula a variância.
# Lembre-se, a unidade aqui seria "notas ao quadrado", o que é difícil de interpretar.

variancia = dados_notas['nota_final'].var()
print(f"A VARIÂNCIA das notas é: {variancia:.2f} (notas2)")
# --- Calculando o DESVIO PADRÃO ---
# O método .std() (Standard Deviation) calcula o desvio padrão.
# Esta é a medida mais importante de dispersão!
desvio_padrao = dados_notas['nota_final'].std()
media_notas = dados_notas['nota_final'].mean()
print(f"O DESVIO PADRÃO das notas é: {desvio_padrao:.2f}")
print(f"\nINTERPRETAÇÃO: Em média, uma nota qualquer da turma tende a estar a {desvio_padrao:.2f} pontos de distância da média ({media_notas:.2f}).")
print("\n" + "="*40 + "\n")
```

```python
# =============================================================================
# AULA 2 - CÓDIGO 2: COMPARANDO A CONSISTÊNCIA DE DUAS TURMAS
# =============================================================================
# Vamos adicionar uma nova turma que foi muito mais consistente (notas próximas).
dados_notas['nota_turma_consistente'] = [7.0, 7.5, 6.5, 7.0, 7.2, 6.8, 7.1, 6.9, 7.3, 7.0]
print("--- Tabela com as Duas Turmas ---")
print(dados_notas)
print("\n" + "-"*40 + "\n")
# Calculando média e desvio padrão para a TURMA ORIGINAL
media_original = dados_notas['nota_final'].mean()
std_original = dados_notas['nota_final'].std()
# Calculando média e desvio padrão para a TURMA CONSISTENTE
media_consistente = dados_notas['nota_turma_consistente'].mean()
std_consistente = dados_notas['nota_turma_consistente'].std()
print(f"--- Turma Original (Inconsistente) ---")
print(f"Média: {media_original:.2f}")
print(f"Desvio Padrão: {std_original:.2f}")
print("\n")
print(f"--- Turma Nova (Consistente) ---")
print(f"Média: {media_consistente:.2f}")
print(f"Desvio Padrão: {std_consistente:.2f}")
print("\n>>> OBSERVE: As médias são quase idênticas! Mas o desvio padrão da turma consistente é MUITO menor,mostrando que as notas estão mais agrupadas e o desempenho é mais previsível.")
```

```python
# =============================================================================
# AULA 2 - EXERCÍCIO: CONTROLE DE QUALIDADE
# =============================================================================
import pandas as pd
# Dados de comprimento (em mm) de parafusos produzidos por duas máquinas.
# O alvo é 50mm.
dados_maquinas = pd.DataFrame({
'maquina_A': [50.1, 50.0, 49.9, 50.0, 49.8, 50.2, 50.0, 50.1, 49.9, 50.0],
'maquina_B': [52.5, 47.0, 50.0, 51.5, 48.0, 46.5, 53.0, 50.5, 49.0, 52.0]
})
print("--- Medidas dos Parafusos (em mm) ---")
print(dados_maquinas)

# Tarefa 1: Calcule a média de produção de cada máquina.
media_maquina_A = dados_maquinas['maquina_A'].mean()
media_maquina_B = dados_maquinas['maquina_B'].mean()
print("\n---- Média de Produção ---")
print(f"Média Máquina A: {media_maquina_A:.2f} mm")
print(f"Média Máquina B: {media_maquina_B:.2f} mm")
print("\n")

# Tarefa 2: Calcule o desvio padrão de cada máquina.
std_maquina_A = dados_maquinas['maquina_A'].std()
std_maquina_B = dados_maquinas['maquina_B'].std()
print("\n--- Desvio Padrão ---")
print(f"Desvio Padrão Máquina A: {std_maquina_A:.2f} mm")
print(f"Desvio Padrão Máquina B: {std_maquina_B:.2f} mm")


# Tarefa 3: Qual máquina é mais confiável e consistente? Justifique sua resposta usando o desvio padrão.
print("\n>>>>> Justificativa - A máquina A é a mais confiável e consistente, porque apesar de ambas apresentarem médias similares, a máquina A apresentou uma produção mais consistente desvio padrão ficou em 0.12mm, isto quer dizer que os parafusos estão com os comprimentos mais proximos do padrão.")
```

```python
#5. Crie um DataFrame com 5 alunos e suas notas finais.
#Calcule:
#Média (.mean())
#Mediana (.median())
#Moda (.mode()[0])
import pandas as pd
dados_nota = pd.DataFrame({
'alunos': ['Amaro', 'João', 'José', 'Maria', 'Vicentina'],
'nota_final': [7.7, 9.0, 9.9, 9.9, 5.5]
})

print("--- Tabela de Notas dos Alunos ---")
print(dados_nota)
print("\n" + "="*30 + "\n")
media = dados_nota['nota_final'].mean()
print(f" Média Nota: {media:.2f}")
mediana = dados_nota['nota_final'].median()
print(f" Mediana Nota: {mediana:.2f}")
Moda = dados_nota['nota_final'].mode() [0]
print(f' Moda Nota: {Moda:.2f}')

```

```python
#8. Crie um DataFrame com notas de 10 alunos.
#Calcule:
#o valor máximo e mínimo,
#a amplitude (diferença entre eles),
#a variância (.var()),
#e o desvio padrão (.std()).
import pandas as pd
dados_notas = pd.DataFrame({
'alunos' : ['Anna', 'Beto', 'Cadu', 'Delta', 'Elma', 'Fabbi', "Gutto", "Hermano", 'Ilca', 'Lucca'],
'notas' : [10.0 , 7.5 , 8.0 , 4.0 , 8.0 , 9.5, 9.0 , 5.0 , 6.5 , 8.0]
})
print(" ---- Tabela de Notas dos Alunos ----")
print(dados_notas)
print(f"\n" + '_'*30 +"\n")
#valor máximo e mínimo
valor_maximo = dados_notas ['notas'].max()
print(f' Nota Máxima: {valor_maximo}')
valor_minimo = dados_notas ['notas'].min()
print(f' Nota Mínima: {valor_minimo}')
#Amplitude
amplitude = (valor_maximo - valor_minimo)
print(f' Amplitude das Notas é: {amplitude:.2f}')
#Variâmcia
variancia = dados_notas['notas'].var()
print(f' A Variância das Notas é: {variancia:.2f}')
#Desvio Padrão
desvio_padrao = dados_notas['notas'].std()
print(f' O Desvio Padrão das Notas é: {desvio_padrao:.2f}')
media = dados_notas ['notas'].mean()
print(f' A Medias entre as Notas é: {media:.2f}')
```
